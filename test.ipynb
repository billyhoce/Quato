### Upload Zipline Strategy to Github

```python
import requests
import dotenv
import base64
import pathlib
import os

dotenv.load_dotenv()

REPO_PATH = "billyhoce/Quato-zipline-strategies/contents"
STRATEGY_FILENAME = "zipline_sample.py"
file = pathlib.Path("./models/zipline_sample.py")

# Read and base64-encode file content
with open(file, "rb") as f:
    content_b64 = base64.b64encode(f.read()).decode("utf-8")

requests.put(
    f"https://api.github.com/repos/{REPO_PATH}/{STRATEGY_FILENAME}",
    headers={
        "Authorization": f"BEARER {os.getenv('GITHUB_TOKEN')}",
        "Accept": "application/vnd.github+json",
        "X-GitHub-Api-Version": "2022-11-28"
    },
    json={
        "message": "Add sample zipline strategy",
        "content": content_b64
    }
)
```
